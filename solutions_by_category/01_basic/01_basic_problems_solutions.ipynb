## 2. Input Age

Create a program that asks the user to enter their name and age. Print out a message addressed to them that tells them the year that they will turn 100 years old.

```python
# Solution:
current_year = 2020
name, age = input('Enter your name and age seperated by space: ').split()
print('{} you will turn into 100 years old in the year {}. Congratualtions!'.format(name, (current_year - int(age))+100))
```

## 3. Calculate with Input

Write a Python program that accepts an integer (n) and computes the value of n+nn+nnn. 
Sample value of n is 5
Expected Result : 615

```python
# Solution:
n = int(input())
n1 = int('%s' % n)
n2 = int('%s%s' % (n,n))
n3 = int('%s%s%s' % (n,n,n))
print(n1 + n2 + n3)
```

## 4. Current Date and Time

```python
# Solution:
import datetime
now = datetime.datetime.now()
print("Current date and time :")
print(now.strftime("%Y-%m-%d %H:%M:%S"))
```

## 5. Circle Area

```python
# Solution:
from math import pi
r = float(input("Input the radius of the circle : "))
print("The area of the circle with radius {} is: {:.2f}".format(r, pi * r**2))
```

## 6. Reverse Name
